<h3>Overview of the Project</h3>

This project examines records and its respective metadata within the first 480 passenger records from the National Archives dataset: Records of Passengers from the Russian Empire, Austria-Hungary, the Netherlands, Scandinavia, and Other European Countries, ca. 1820 - ca. 1912 (titled PAS_RUS). The PAS_RS is one of 4 subset databases of the MS Access database donated to the National Archives, with the three other databases covering Italian passenger records (PAS_IT), UK and Irish passenger records (PAS_UK), and German, French, and Switzerland passenger records (PAS_GS). PAS_RS has the smallest amount of passenger records, containing 835,984 records total, while the three other databases contain two million or more.

The weblink within the National Archives Catalog also contains a readme document detailing the organization and column names/categories within the databases. The established structure includes name, age, marital status, location of origin, destination, literacy, family relationships within the US, phenotypes, and more. However, as you'll read further in this notebook, most of this information is excluded in the raw PAS_RS database leading to major data cleaning and minor research. The project aims to refine the dataset's missing information and make inferences based on existing data including the ages, sex, country of origin, and city of destination of the passenger records. 

The project uses OpenRefine and Tableau to refine, filter, and create data visualizations of the records. It also makes interpretations of the data using the readme file provided by the National Archives, NotebookLM to clarify codes (but not make any data modifications/refinement), and Google search to look up town names. This notebook can hopefully be used as an outline to perform similar modifications and analysis to the entirety of the database using a more powerful local machine.

The PAS_RS raw data and Readme documentation are attached to this folder and at the bottom of this notebook.
