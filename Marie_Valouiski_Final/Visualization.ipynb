## Data Visualization

### Tableau

I used Tableau, which is a free to use, "fast analytics and rapid-fire business intelligence" software (with paywalled features) to create my data visualizations mostly measuring count of records against categories and concentrations of people. This helped visualize the relationship between passenger ship records, their countries of origin, demographics, and the city of destination within the US.

I made a total of 8 charts within a Story, a feature within Tableau to help display computational stories with data. Here are each of the charts and their findings:

**Regions and literacy rates of passenger records.**

Stacked bar chart showing if passengers are literate (assuming in English) shown as Yes, No, or Unknown based on Region (Russian empire or Scandinavian). Of the 480 records, only 15 records are listed as Yes and 7 records are listed as No within the Russian empire, while the rest are listed as Unknown. This visualization mainly showcases the uneven distribution of regions within the selected dataset, with "Russian empire" encompassing countries such as Finland and Germany according to the PAS_RUS dataset.



**Country of origin and count of passengers**

Horizontal bar chart displaying the distribution of passengers based on the country column that I created. Again, most of the countries of origin are listed as unknown, however we can see that there is also a substantial amount of records from countries like Germany (80), Poland (44), and Russia (30) from the first 480 records. It also shows that of the 480 records, 30 are from passengers who are returning to the USA.



**Geographical visualization of passenger count and country of origin**

This mainly displays the same information as the previous chart, but in a more visual and interactive method. Tableau is able to automatically recognize countries within a dataset and distribute the datapoints within a map to show the concentration of passengers within a country of origin. Here we can see the darker blue countries contain more records, while lighter countries contain less. Below the chart, Tableau lets us know there are 9 records with unknown locations, which could be attributed to the datapoints in which I could not find an associated country and the original location was left unmodified.



**Age groups and count of passengers**

Pie chart showing the age groups based on the Age_range column I created and their composition of the 480 total records. We can see that almost three quarters of the dataset is composed of passengers aged 18 or older, however there is also a subtantial amount of children that accompanied, most likely, their parents. Note that hovering over the red shows that there are 22 infants within the dataset, however the chart didn't label it for some reason.



**Composition of sex within age groups**

Stacked bar chart that reveals sex composition within the established age groups. It conveys that within adults, males outweigh females almost 2:1, however within children and infants the groups are almost exactly evenly distributed (for clarification within infants, there are 10 females and 10 males of the 20 infants within the 480 records).



**Destination location distribution of passengers**

Packed bubble chart showing the destination locations of the passengers and their respective quantities. We can see that the overwhelming majority of passengers had New York as their destination location, but we also see some distribution within other countries/states such as Canada (6), Pennsylvania (which included Philadelphia separately), IL which can be interpreted as an abbreviation of Illinois, MN which can be interpreted as Minnesota (3), and CA which can be interpreted as California (2). We also see other destination cities outside of these territories inluding Baltimore (7, also 1 other in Maryland in Annapolis) and Boston (2).
